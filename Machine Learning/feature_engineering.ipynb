```python
import pandas as pd
import numpy as np
```

Feature Creation

 Financial Data: Price-to-Earnings (P/E) Ratio

```python
data={
    'stock_price':[100,120,140,160,180],
    'earnings_per_share':[5,6,7,8,9]
}
df=pd.DataFrame(data)
print(df)
```

```python
#calculate p/e (price to earning) ratio by dividing stock price by earnings per share
df['P/E Ratio']=df['stock_price'] / df['earnings_per_share']
print(df)
```

Healthcare Data: Body Mass Index (BMI)

```python
data={
    'weight':[60,70,80,90,100],
    'height':[1.60,1.70,1.75,1.80,1.85]
}
df=pd.DataFrame(data)
print(df)
```

```python
#calculate bmi
df['bmi']=df['weight'] / df['height'] ** 2
print(df)
```

```python
#add bmi category
def bmi_category(bmi):
    if bmi < 18.5:
        return 'Under Weight'
    elif 18.5 <= bmi < 24.5:
        return 'Normal Weight'
    elif 25 <= bmi < 29.9:
        return 'Over Weight'
    else:
        return 'Obese'
    
```

```python
df['category']=df['bmi'].apply(bmi_category)
print(df)
```

 Retail Data: Day of the Week (for Sales Prediction)

```python
data={
    'timestamp':['2023-01-01 12:00:00','2023-02-01 14:30:00','2023-03-01 16:45:00']
} 
df=pd.DataFrame(data)
print(df)
```

```python
df['timestamp']=pd.to_datetime(df['timestamp'])
print(df)
```

```python
# Extract datetime features
df['year']=df['timestamp'].dt.year
df['month']=df['timestamp'].dt.month
df['day']=df['timestamp'].dt.day
df['hour']=df['timestamp'].dt.hour
df['minute']=df['timestamp'].dt.minute
df['weekday']=df['timestamp'].dt.weekday
print(df)
```

 E-commerce Data: Price Range

```python
data={
    'product':['A','B','C','D','E'],
    'price':[15,100,250,50,300]
}
df=pd.DataFrame(data)
print(df)
```

```python
# Define price categories (cheap,medium,expensive)
def price_categories(price):
    if price < 50:
        return 'Cheap'
    elif 50 <= price < 200:
        return 'Medium'
    else:
        return 'Expensive'
```

```python
df['Price_Range']=df['price'].apply(price_categories)
print(df)
```

Weather Data: Temperature Ranges

```python
data={
    'Date':['2025-04-01','2025-04-02','2025-04-03','2025-04-04'],
    'Temperature':[25,15,30,10]
}
df=pd.DataFrame(data)
print(df)
```

```python
def temperature_range(temp):
    if temp < 15:
        return 'Cold'
    elif 15 <= temp < 25:
        return 'Warm'
    else:
        return 'Hot'
```

```python
df['Temperature Range']=df['Temperature'].apply(temperature_range)
print(df)
```

Feature Transformations

1. Log Transformations

```python
data={
    'value':[1,4,9,16,25]
}
df=pd.DataFrame(data)
print(df)
```

```python
df['log_value']=np.log(df['value'])
print(df)
```

```python
# If the data contains zero' pr negative values, add a constant (e.g.,1) befor apply!ng log.
#this avoid issues with undefined log values like log(0) or negative
df['log_value_safe']=np.log(df['value']+1)
print(df)
```

2. Custom Transformations

```python
#Binning
data={
    'age':[12,35,45,23,67,89]
}
df=pd.DataFrame(data)
print(df)
```

```python
# Define bins
bins=[0,18,35,55,100]
labels=['0-18','19-35','36-55','56+']
```

```python
df['age_group']=pd.cut(df['age'], bins=bins, labels=labels)
print(df)
```

3. Categorical Encoding

3. 1. One-Hot Encoding 

```python
data={
    'category':['A','B','A','C','B']
}
df=pd.DataFrame(data)
print(df)
```

```python
# Apply One-Hot Encoding
df=pd.get_dummies(df,columns=['category'], dtype=int)
print(df)
```

3. 2. Label Encoding 

```python
data={
    'category':['cat','dog','dog','cat','rabbit','rabbit']
}
df=pd.DataFrame(data)
print(df)
```

```python
df['category_encoded']=pd.factorize(df['category'])[0]
print(df)
```
