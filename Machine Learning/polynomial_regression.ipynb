```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Datasets/Polynomial_Regression_Position_Salaries.csv')
df.head()
```

```python
X=df.iloc[:,1:2].values
y=df.iloc[:,2].values
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score
```

```python
model1=LinearRegression()
```

```python
model1.fit(X,y)
```

```python
y_pred1=model1.predict(X)
```

```python
mse1=mean_squared_error(y_pred1,y)
mae1=mean_absolute_error(y_pred1,y)
rmse1=root_mean_squared_error(y_pred1,y)
r2score1=r2_score(y_pred1,y)

print("MAE: ",mae1)
print("MSE: ",mse1)
print("RMSE: ",rmse1)
print("r2_score: ",r2score1)
```

```python
plt.scatter(X,y,color="blue") 
plt.plot(X, model1.predict(X),color="red")
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly_regs=PolynomialFeatures(degree=5)
```

```python
X_poly=poly_regs.fit_transform(X)
x_train,x_test,y_train,y_test=train_test_split(X_poly,y,test_size=0.2,random_state=0)
```

```python
model2=LinearRegression()
```

```python
model2.fit(x_train,y_train)
```

```python
y_pred2=model2.predict(x_test)
```

```python
mse2=mean_squared_error(y_pred2,y_test)
mae2=mean_absolute_error(y_pred2,y_test)
rmse2=root_mean_squared_error(y_pred2,y_test)
r2score2=r2_score(y_pred2,y_test)

print("MAE: ",mae2)
print("MSE: ",mse2)
print("RMSE: ",rmse2)
print("r2_score: ",r2score2)
```

```python
plt.scatter(X,y,color="blue") 
plt.plot(X, model2.predict(poly_regs.fit_transform(X)),color="red")
```
