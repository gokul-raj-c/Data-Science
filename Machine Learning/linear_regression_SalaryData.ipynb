```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, root_mean_squared_error
```

Load Dataset

```python
df=pd.read_csv('Datasets/Linear_Regression_Salary_Data.csv')
df
```

```python
#X=df[['YearsExperience']] 
#print(X.shape)# Xshould be two dimensional 
#y=df['Salary'] 
```

```python
X= df.iloc[:, :-1].values  # full rows(:) from 0th column(:1) years of Experience 
print(X.shape)# Xshould be two dimensional 
y = df.iloc[:, 1].values # Entire rows from second column(Salary)
```

```python
x_train,x_test,y_train,y_test=train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
#Prediction result on Train Data 
y_pred1= model.predict(x_train) 
```

```python
#Prediction result on Test Data 
y_pred2 = model.predict(x_test) 
```

```python
plt.scatter(x_train, y_train, color="green") 
plt.plot(x_train, y_pred1, color="red") 
plt.title("Salary vs Experience (Training Dataset)") 
plt.xlabel("Years of Experience") 
plt.ylabel("Salary(In Rupees)") 
plt.show()
```

```python
#visualizing the Test set results 
plt.scatter(x_test, y_test, color="blue") 
plt.plot(x_test, y_pred2, color="red") 
plt.title("Salary vs Experience (Test Dataset)") 
plt.xlabel("Years of Experience") 
plt.ylabel("Salary(In Rupees)") 
plt.show() 
```

```python
#Visualizing the Test set results
print("Mean:",df['Salary'].mean()) 
```

```python
mae=mean_absolute_error(y_test,y_pred2)
mse=mean_squared_error(y_test,y_pred2)
r2score=r2_score(y_test,y_pred2)
print("MAE: ",mae)
print("MSE: ",mse)
print("r2_score: ",r2score)
```

```python
#rmse = np.sqrt(mean_squared_error(y_test, y_pred))
rmse=np.sqrt(mse)
print(rmse)
```

```python
mean_value = y_test.mean()
print(mean_value)

if rmse < 0.10 * mean_value:
    print("RMSE is less than 10% of mean")
else:
    print("RMSE is greater than or equal to 10% of mean")
```

```python
rmse_percentage = (rmse / mean_value) * 100

print("RMSE Percentage:", rmse_percentage, "%")
```
