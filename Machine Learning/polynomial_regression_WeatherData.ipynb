```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Datasets/Polynomial_Regression_WeatherDataP.csv')
df_copy=df.copy()
df_copy.head()
```

```python
df_copy=df_copy.rename(columns={
    'Pressure (millibars)':'Pressure'
})
```

```python
df_copy
```

```python
X=df_copy.iloc[:,0:1].values
y=df_copy.iloc[:,1].values
```

```python
plt.scatter(X,y,color="blue")
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly_regs=PolynomialFeatures(degree=10)
```

```python
X_poly_deg2=poly_regs.fit_transform(X)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score
```

```python
model1=LinearRegression()
```

```python
model1.fit(X_poly_deg2,y)
```

```python
plt.scatter(X,y,color="blue") 
plt.plot(X, model1.predict(X_poly_deg2), color="red")
```
