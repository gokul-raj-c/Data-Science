Missing values: count, percentage, and appropriate treatment.


```python
# Missing value count
missing_count = df.isnull().sum()
print(missing_count)
```

```python
cols = [
    "delivery_days",
    "estimated_delivery_days",
    "return_status",
    "return_reason",
    "customer_rating",
    "review_sentiment"
]

for col in cols:
    print(f"\n--- {col} ---")

    result = pd.crosstab(
        df["order_status"],
        df[col].isna().map({
            True: "Missing",
            False: "Present"
        })
    )

    print(result)
```

```python
# Delivery analysis
completed_df = df[df["order_status"] == "Completed"].copy()

# Return reason analysis
returned_df = df[df["order_status"] == "Returned"].copy()

print("Completed orders:", len(completed_df))
print("Returned orders:", len(returned_df))
```

```python
print("Completed orders:")
print(completed_df[
    ["delivery_days", "estimated_delivery_days",
     "customer_rating", "review_sentiment"]
].isnull().sum())

print("\nReturned orders:")
print(returned_df[
    ["return_status", "return_reason"]
].isnull().sum())
```

The missing values in delivery_days, estimated_delivery_days, customer_rating, review_sentiment, return_status, and return_reason are strongly associated with order_status. Delivery information and customer feedback are available only for completed orders, while return information is available only for returned orders. Cancelled and pending orders have missing values in all six columns.

This pattern suggests that the missing values are structural rather than random, as the information is recorded only for specific order statuses. Therefore, these null values should not be treated as data errors or filled with arbitrary values. The columns should be retained, and the relevant subsets of orders should be used for each analysis. This approach preserves the dataset's information while ensuring meaningful and accurate EDA results.

```python
# Check coupon code availability against discount amount
df["coupon_status"] = df["coupon_code"].isna().map({
    True: "No coupon code recorded",
    False: "Coupon code recorded"
})

df["discount_status"] = df["discount_amount"].apply(
    lambda x: "Zero discount" if x == 0
    else "Discount applied" if x > 0
    else "Negative/invalid discount"
)

print(pd.crosstab(
    df["coupon_status"],
    df["discount_status"]
))
```

```python
df = df.drop(columns=["campaign_name", "coupon_code"])
```

```python

# Calculate missing-value statistics
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})

# Keep columns that contain missing values
missing_summary = missing_summary[
    missing_summary["Missing Count"] > 0
].copy()

# Recommended treatment based on your investigation
treatment = {
    "delivery_days":
        "Keep; analyze completed orders only",
    "estimated_delivery_days":
        "Keep; analyze completed orders only",
    "customer_rating":
        "Keep; analyze completed orders only",
    "review_sentiment":
        "Keep; analyze completed orders only",
    "return_status":
        "Keep; analyze returned orders only",
    "return_reason":
        "Keep; analyze returned orders only"
}

missing_summary["Recommended Treatment"] = [
    treatment.get(
        col,
        "Investigate missingness before deciding treatment"
    )
    for col in missing_summary.index
]

print(missing_summary.to_string())

# Save the report
missing_summary.to_csv("missing_values_report.csv")

print("\nMissing-value report saved successfully.")
```

Validate financial calculations, statuses, and currency consistency.


```python
# Check financial columns
financial_cols = [
    "gross_sales", "discount_amount", "tax_amount",
    "shipping_cost", "net_sales", "product_cost",
    "profit", "profit_margin_percentage"
]

print(df[financial_cols].describe())
```

```python

# 1. Check net sales against gross sales minus discount
df["expected_net_sales"] = (
    df["gross_sales"] - df["discount_amount"]
)

df["net_sales_difference"] = (
    df["net_sales"] - df["expected_net_sales"]
)

print("Net sales formula check:")
print("Matching rows:",
      df["net_sales_difference"].abs().lt(0.02).sum())
print("Non-matching rows:",
      df["net_sales_difference"].abs().ge(0.02).sum())

# 2. Check profit margin against profit / net sales
df["calculated_margin"] = (
    df["profit"] / df["net_sales"] * 100
)

df["margin_difference"] = (
    df["profit_margin_percentage"] - df["calculated_margin"]
)

print("\nProfit margin formula check:")
print("Matching rows:",
      df["margin_difference"].abs().lt(0.02).sum())
print("Non-matching rows:",
      df["margin_difference"].abs().ge(0.02).sum())

# 3. Inspect examples that do not match
print("\nNet sales mismatches:")
print(df.loc[
    df["net_sales_difference"].abs().ge(0.02),
    ["gross_sales", "discount_amount", "tax_amount",
     "shipping_cost", "net_sales", "net_sales_difference"]
].head())

print("\nProfit margin mismatches:")
print(df.loc[
    df["margin_difference"].abs().ge(0.02),
    ["net_sales", "profit", "profit_margin_percentage",
     "calculated_margin"]
].head())
```

```python

# Calculate expected net sales
df["expected_net_sales"] = (
    df["gross_sales"]
    - df["discount_amount"]
    + df["tax_amount"]
    + df["shipping_cost"]
)

# Compare with recorded net sales
df["net_sales_difference"] = (
    df["net_sales"] - df["expected_net_sales"]
).round(2)

tolerance = 0.02

matching = df["net_sales_difference"].abs() < tolerance

print("Matching rows:", matching.sum())
print("Non-matching rows:", (~matching).sum())

print("\nExamples of mismatches:")
print(
    df.loc[~matching, [
        "gross_sales", "discount_amount", "tax_amount",
        "shipping_cost", "net_sales", "expected_net_sales",
        "net_sales_difference"
    ]].head()
)
```

Financial calculations were validated across all 138,116 records. Net sales matched the formula gross sales minus discount amount plus tax and shipping costs for every record, with zero mismatches. Similarly, profit margin percentage matched profit divided by net sales multiplied by 100 for all records. These results confirm that the two tested financial calculations are internally consistent with the dataset. Therefore, no corrections are required for these calculations before proceeding with the EDA.

3. Data Exploration 


1. Which day of the week generates the most orders?

```python
# Define weekday order
weekday_order = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

# Count unique orders by weekday
weekday_orders = (
    df.dropna(subset=["order_date"])
      .assign(weekday=lambda x: x["order_date"].dt.day_name())
      .groupby("weekday")["order_id"]
      .nunique()
      .reindex(weekday_order, fill_value=0)
      .reset_index(name="order_count")
)

display(weekday_orders)

# Identify the day with the most orders
print("Day with the most orders:")
display(weekday_orders.loc[weekday_orders["order_count"].idxmax()])
```

```python
plt.figure(figsize=(10, 5))

bars = plt.bar(
    weekday_orders["weekday"],
    weekday_orders["order_count"]
)

plt.title("Number of Orders by Day of the Week")
plt.xlabel("Day of the Week")
plt.ylabel("Number of Orders")
plt.xticks(rotation=30)
plt.bar_label(bars, fmt="%.0f", padding=3)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

 Saturday recorded the highest number of orders (19,950), whereas Sunday recorded the lowest (19,480). However, the difference between these days was relatively small, at only 470 orders. Overall, order volumes remained fairly consistent throughout the week, indicating no pronounced day-of-week pattern. Higher Saturday activity may be associated with weekend shopping behavior

2. Which months have the highest and lowest order volumes?

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

monthly_orders = (
    df.dropna(subset=["order_date"])
      .assign(month=df["order_date"].dt.month_name())
      .groupby("month")["order_id"]
      .nunique()
      .reindex(month_order, fill_value=0)
      .reset_index(name="order_count")
)

display(monthly_orders)

# Highest and lowest order-volume months
print("Month with highest order volume:")
display(monthly_orders.loc[monthly_orders["order_count"].idxmax()])

print("Month with lowest order volume:")
display(monthly_orders.loc[monthly_orders["order_count"].idxmin()])
```

```python
plt.figure(figsize=(12, 6))

bars = plt.bar(
    monthly_orders["month"],
    monthly_orders["order_count"]
)

plt.title("Order Volume by Calendar Month (2021–2025)")
plt.xlabel("Month")
plt.ylabel("Total Number of Orders")
plt.xticks(rotation=45)
plt.bar_label(bars, fmt="%.0f", padding=3, fontsize=8)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

December recorded the highest number of orders (17,567), followed by November (16,061), while February recorded the lowest (8,027). Order volumes were generally higher toward the end of the year, suggesting a possible seasonal increase in customer purchasing activity. Holiday shopping, year-end promotions, and major sales campaigns may contribute to this pattern

3. Which regions generate the highest net sales?	


```python

# Net sales by region and currency
regional_sales = (
    df.dropna(subset=["region", "net_sales", "currency"])
      .groupby(["region", "currency"], observed=True)["net_sales"]
      .sum()
      .reset_index()
      .sort_values(["currency", "net_sales"], ascending=[True, False])
)

display(regional_sales)
```

```python

top_regions = (
    regional_sales
    .loc[regional_sales.groupby("currency")["net_sales"].idxmax()]
    .sort_values("currency")
)

display(top_regions)
```

```python

import math

currencies = sorted(regional_sales["currency"].unique())

ncols = 2
nrows = math.ceil(len(currencies) / ncols)

fig, axes = plt.subplots(
    nrows, ncols,
    figsize=(15, 4 * nrows),
    squeeze=False
)

for ax, currency in zip(axes.flatten(), currencies):
    data = (
        regional_sales[regional_sales["currency"] == currency]
        .sort_values("net_sales", ascending=True)
    )

    ax.barh(data["region"], data["net_sales"])
    ax.set_title(f"Regional Net Sales ({currency})")
    ax.set_xlabel(f"Net Sales ({currency})")
    ax.set_ylabel("Region")
    ax.grid(axis="x", alpha=0.3)

# Hide unused subplot spaces
for ax in axes.flatten()[len(currencies):]:
    ax.set_visible(False)

plt.suptitle("Net Sales by Region and Currency", fontsize=15)
plt.tight_layout()
plt.show()
```
