```python
import pandas as pd
```

```python
df=pd.read_csv('G.csv')
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df=df.dropna()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.columns = df.columns.str.strip().str.replace(' ', '_')
```

```python
df.info()
```

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'])
```

```python
df['Ship_Date']=pd.to_datetime(df['Ship_Date'])
```

```python
df.info()
```

```python
df['Profit'] = df['Profit'].str.replace('$', '', regex=False)
```

```python
df['Profit']
```

```python
df["Profit"] = pd.to_numeric(df["Profit"]).astype(int)
```

```python
df.info()
```

```python
df1=df.groupby('Segment')['Profit'].sum().reset_index()
```

```python
df1
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.barplot(x='Segment', y='Profit', data=df1)
plt.plot()
```
