```python
import numpy as np
```

Joining Arrays


1. numpy.concatenate

```python
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
print(a)
print()
print(b)
```

```python
print(np.concatenate((a,b)))
```

```python
print(np.concatenate((a,b),axis=1))
```

2. numpy.hstack

```python
print(np.hstack((a,b)))
```

3. numpy.vstack

```python
print(np.vstack((a,b)))
```

Array Manipulation

1. numpy.reshape

```python
a=np.array([1,2,3,4,5,6])
print(a)
print()
print(a.reshape(2,3))
print()
print(a.reshape(3,2))
```

2. numpy.ndarray.flatten

```python
a = np.arange(8).reshape(2,4)
print(a)
print()
print(a.flatten())
print()
print(a.flatten(order='F'))
```

3. numpy.ndarray.flat

```python
a=np.arange(8).reshape(2,4)
print(a)
print()
# returns element corresponding to index
print(a.flat[6])
```

4. Splitting Arrays

```python
a = np.arange(9)
print(a)
```

```python
#equal 3 sizes
b=np.split(a,3)
print(b)
```

```python
b=np.split(a,[4,7])
print(b)
```

5. numpy.resize

```python
a=np.array([[1,2,3],[4,5,6]])
print(a)
print(a.shape)
```

```python
a1=np.resize(a,(3,2))
print(a1)
```

```python
a2=np.resize(a,(3,3))
print(a2)
```

This function returns a new array with the specified size. If the new size is
greater than the original, the repeated copies of entries in the original are
contained.

6. numpy.append

This function adds values at the end of an input array.

```python
a=np.array([[1,2,3],[4,5,6]])
print(a)
b=np.array([7,8,9])
print(b)
```

```python
print(np.append(a,b))
```

```python
print(np.append(a,[b],axis=0))
```

```python
print(np.append(a,[[7,8],[9,10]],axis=1))
```

7. numpy.insert

```python
arr=np.array([[1,2],[3,4],[5,6]])
print(arr)
```

```python
print(np.insert(arr, 3, [11, 12]))
```

```python
print(np.insert(arr, 1, [11],axis=0))
```

```python
print(np.insert(arr, 1, [11,12],axis=0))
```

```python
print(np.insert(arr, 1, [11],axis=1))
```

8. numpy.delete

```python
a = np.arange(12).reshape(3,4)
print(a)
```

```python
print(np.delete(a, 5))
```

```python
print(np.delete(a, 1,axis=1))
```

```python
print(np.delete(a, 1,axis=0))
```

9. numpy.unique

This function returns an array of unique elements in the input array.


```python
a=np.array([5,2,6,7,7,2,8,2,9])
print(a)
```

```python
print(np.unique(a))
```

```python
#position
u,indices=np.unique(a,return_index=True)
print(u)
print(indices)
```

```python
#count
u,indices=np.unique(a,return_counts=True)
print(u)
print(indices)
```
