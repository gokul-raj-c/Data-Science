6. numpy.linalg.solve()

3x+4Y+5z=9
y+2X+3z=8
2x+4y+5z=7

```python
a=np.array([[3,4,5],[1,2,3],[2,4,5]])
b=np.array([9,8,7])
ans=np.linalg.solve(a,b)
print(ans)
```

```python
# Checking if the ånywer if correct
print(np.allclose(np.dot(a, ans),b))
```

7. numpy.linalg.inv()

We use numpy.linalg.inv() function to calculate the inverse of a matrix.

```python
x=np.array([[1,2],[3,4]])
y = np.linalg.inv(x)
print(x)
print(y)
```

```python
print(np.dot(x, y))
```
