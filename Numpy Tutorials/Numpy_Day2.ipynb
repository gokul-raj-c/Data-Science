Broadcasting

```python
a=np.array([1,2,3,4])
b=np.array([10,20,30,40])
c=a*b
print(c)
```

```python
a=np.arange(0,16).reshape(4,4)
print(a)
print()
b=np.array([1,2,3,4])
print(b)
print()
print(a+b)
```

Functions for Rounding

1. numpy.around()

```python
a = np.array([1.0,5.55, 123, 0.567, 25.532])
print(a)
print()
print(np.round(a))
print()
print(np.round(a,decimals=1))
```

2. numpy.ceil()

```python
a = np.array([-1.7, 1.5, -0.2, 0.6, 10])
print(a)
print()
print(np.ceil(a))
```

3. numpy.floor()


```python
print(np.floor(a))
```

Arithmetic Operations

```python
a = np.array([[10,20,30],[40,50,60],[70,80,90]])
print(a)
print()
b = np.array([10,10,10])
print(b)
```

```python
print(np.add(a, b))
```

```python
print(np.subtract(a, b))
```

```python
print(np.multiply(a, b))
```

```python
print(np.divide(a, b))
```

4. numpy.reciprocal()

```python
a = np.array([0.25, 1.33, 1, 0, 100])
print(a)
print()
print(np.reciprocal(a))
```

5. numpy.power()

```python
a = np.array([10,100,1000])
print(a)
print()
print(np.power(a, 2))
```

6. numpy.mod()

```python
a = np.array([10,20,30])
b = np.array([3,5,7])
print(a)
print()
print(b)
print()
print(np.mod(a, b))
```

Iterating Over Array

```python
a = np.array([[0,5,10,15],[20,25,30,35],[40,45,50,55]])
print(a)
```

```python
for x in np.nditer(a):
  print(x, end=' ')
```

```python
c = a.copy(order = 'C')
print(c)
for x in np.nditer(c):
  print(x, end=' ')
```

```python
c = a.copy(order = 'F')
print(c)
for x in np.nditer(c):
  print(x, end=' ')
```
