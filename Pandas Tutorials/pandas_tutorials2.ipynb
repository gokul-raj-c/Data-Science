```python
import pandas as pd
import numpy as np
```

Filter Pandas Dataframe with multiple conditions

```python
data={
    'Name':['RACHEL', 'MONICA', 'PHOEBE', 'ROSS','CHANDLER','JOEY'],
    'Age': [30,35,37,33,34,30],
    'Salary':[100000,93000,88000,120000,94000,95000],
    'Job':['DESIGNER','CHEF','MASUS','PALENTOLOGY','IT','ARTIST']
}
df=pd.DataFrame(data)
print(df)
```

1. Filter Pandas Dataframe with multiple conditions Using loc

```python
df.loc[(df['Salary']>=10000) & (df['Age']<40) & (df['Job'].str.startswith('D')),['Name','Job']]
```

2. Filter Pandas Dataframe Using NumPy

```python
filtered_values=np.where((df['Salary']>=10000) & (df['Age']<40) & (df['Job'].str.startswith('D')))
df.loc[filtered_values]
```

3. Filter Pandas Dataframe Using Query (eval and query works only
with columns)

```python
print(df.query('Salary<=100000 & Age<40 & Job.str.startswith("D").values'))
```

4. Pandas Boolean indexing multiple conditions standard way
("Boolean indexing" works with values in a column only)

```python
df[(df['Salary']>=10000) & (df['Age']<40) & (df['Job'].str.startswith('D'))][['Name','Age','Job']]
```

5. Eval multiple conditions ("eval" and "query" works only with
columns )

```python
df[df.eval('Salary<=100000 & Age<40 & Job.str.startswith("D").values')]
```

eval method can also be used to perform arithmetic operations

```python
df['Salary_in_thousands']=df.eval('Salary/1000')
df
```

```python
df['Age']=df.eval('Age-5')
df
```

6. How to Filter DataFrame by List of Values in Pandas

```python
job_list=['DESIGNER','ARTIST']
df[df['Job'].isin(job_list)]
```

7. How to Filter DataFrame using filter in Pandas

```python
fileterd_df=df.filter(items=['Name','Age'])
print(fileterd_df)
```

8. Selecting Columns usinu filter and like

```python
# Select columns whose names contain 'a'
fileterd_df=df.filter(like='a',axis=1)
print(fileterd_df)
```

```python
df_copy=df.copy()
```

```python
def convert(var):
    if var <= 1000:
        return 'Basic'
    elif var > 1000 and var <= 5000:
        return 'Intermediate'
    elif var > 5000:
        return 'Premium'
```

```python
df_copy['Customer_Type']=df_copy['Sales'].apply(convert)
df_copy
```

```python
df_copy['Customer_Type'].value_counts()
```

```python
df_copy['Sales_round']=df_copy['Sales'].round(1)
df_copy.head()
```
