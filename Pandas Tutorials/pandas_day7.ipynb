```python
import pandas as pd
```

```python
df = pd.read_excel('https://github.com/datagy/mediumdata/raw/master/sample_pivot.xlsx', parse_dates=['Date'] )
```

```python
df.head()
```

Aggregating Only Certain Colums in a Pandas Pivot Table

```python
pivot=pd.pivot_table(data=df, index='Region', values='Sales')
pivot
```

```python
pivot=pd.pivot_table(data=df, index='Region', values=['Sales','Units'])
pivot
```

Working with Aggregation Methods in a Pandas PivotTable

```python
#by default aggfunction is mean
pivot=pd.pivot_table(data=df, index='Region', values='Sales', aggfunc='sum')
pivot
```

Multiple Aggregation Method in a Pandas DataFrame

```python
pivot=pd.pivot_table(data=df, index='Region', values='Sales', aggfunc=['sum','mean'])
pivot
```

Specifying Different Aggregations Per Column

```python
pivot=pd.pivot_table(data=df, index='Region', aggfunc={'Sales':'mean', 'Units':'sum'})
pivot
```

Custom Aggregations in Pandas Pivot Tables

```python
import numpy as np
```

```python
# Defining a custo Onction
def mean_no_outliers(values):
    no_outliers= values.quantile([0.1,0.9])
    mean=np.mean(no_outliers)
    return mean
pivot=pd.pivot_table(data=df, index='Region', values='Sales', aggfunc=['mean', mean_no_outliers])
pivot
```

Add Columns to a Pandas Pivot Table

```python
pivot=pd.pivot_table(data=df, index='Region', columns='Type', values='Sales')
pivot
```

Filter data by quarters (4 quarters in a month)

```python
pivot=pd.pivot_table(data=df, index=['Region', df['Date'].dt.quarter], columns='Type', values='Sales')
pivot
```

Accessing data in a multi-index Pandas DataFrame_

```python
print(pivot.loc[('East',1),"Men's Clothing"])
```

Customizing Pandas Pivot Tables

```python
pivot=pd.pivot_table(data=df, index='Region', columns='Type', values='Sales', margins=True)
pivot
```

 Renaming column as total in a Pandas pivot table

```python
pivot=pd.pivot_table(data=df, index='Region', columns='Type', values='Sales', margins=True, margins_name='Total')
pivot
```

Dealing with Missing Datai a Pandas Pivot Table

Sorting Pandas Pivot Table Data

```python
pivot=pd.pivot_table(data=df, index='Region', values='Sales', sort=True)
pivot
```

Filtering Python Pivot Tables

```python
pivot=pd.pivot_table(data=df, index=['Region', df['Date'].dt.quarter], values='Sales')
print(pivot[pivot['Sales'] > pivot['Sales'].mean()])
```

Dealing with Missing Datai a Pandas Pivot Table

```python
df.loc[(df['Region']=='East') & (df['Type']=="Children's Clothing"), 'Sales']=np.nan
```

```python
pivot=pd.pivot_table(data=df, index='Region', columns='Type', values='Sales')
pivot
```

 Filling Missing Values in a Pandas Pivot Table

```python
pivot=pd.pivot_table(data=df, index='Region', columns='Type', values='Sales', fill_value=0)
pivot
```
