Pandas - Descriptive Statistics

```python
from scipy.stats import skew, kurtosis
```

```python
def descriptive_statistics(df,column):
  mean=df[column].mean()
  median=df[column].median()
  mode=df[column].mode()[0]  # Mode returns a series, get the first element
  variance=df[column].var()
  std_dev=df[column].std()
  data_range=df[column].max()-df[column].min()
  min_value=df[column].min()
  max_value=df[column].max()
  skewness= skew(df[column])
  kurt=kurtosis(df[column], fisher=True)
  percentiles=[
      df[column].quantile(0.25),
      df[column].quantile(0.50),
      df[column].quantile(0.75)
  ]
  statistics={
      'mean':mean,
      'median':median,
      'mode': mode,
      'Variance': variance,
      'Standard Deviation':std_dev,
      'range':data_range,
      'min value':min_value,
      'max value':max_value,
      'Skewness': skewness,
      'Kurrtosis': kurt,
      '25th Percentile':percentiles[0],
      '50th Percentile':percentiles[1],
      '75th Percentile':percentiles[2],
  }
  return pd.Series(statistics)

```

```python
data={
    'name':pd.Series(['tom','james','ricky','vin','steve','smith','jack']),
    'age':pd.Series([24,25,26,22,23,21,20]),
    'rating':pd.Series([4.2,3.4,3.9,2.6,3.2,4.6,3.8])
}
df=pd.DataFrame(data)
```

```python
age_statistics=descriptive_statistics(df,'age')
print(age_statistics)
```

2. Remove Outliers sing IQR (Interquartile Range)

```python
data={'col1':[10,20,30,40,100,200,300,500,1000]}
df=pd.DataFrame(data)
```

```python
q1=df['col1'].quantile(0.25)
q3=df['col1'].quantile(0.75)
iqr=q3-q1
```

```python
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr
```

```python
df_no_outliers=df[(df['col1'] >= lower_bound) & (df['col1'] <= upper_bound)]
print(df_no_outliers)
```

3. Remove Outliers Using modified Z-Score

```python
data={'col1':[10,20,30,40,100,200,300,500,1000]}
df=pd.DataFrame(data)
```

```python
# Calculate the median and MAD
median=np.median(df['col1'])
mad=np.median(np.abs(df['col1']-median))
```

```python
# Calculate the modified Z-score
modified_z_scores = 0.6745 * (df['col1']-median) / mad
```

```python
# Filter out rows with m6dified u-scores greater than 3.5
#(absolute value)
df_no_outliers= df[abs(modified_z_scores) <= 3.5]
print(df_no_outliers)
```
