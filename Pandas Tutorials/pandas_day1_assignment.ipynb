22.Given a DataFrame of students with Name, Marks1, Marks2:
Add a column &quot;Total&quot;
Add a column &quot;Grade&quot; based on total marks
Filter students with Grade &#39;A&#39;

```python
import pandas as pd
data ={
    "Name": ["gokul","adwaith","allen","nithin"],
    "marks1": [85, 70, 95, 60],
    "marks2": [90, 75, 92, 65]
}
df = pd.DataFrame(data)
df["total"] = df["marks1"] + df["marks2"]
print(df)
```

```python
# Add Grade column based on Total
def get_grade(total):
    if total >= 170:
        return "A"
    elif total >= 140:
        return "B"
    elif total >= 100:
        return "C"
    else:
        return "D"

df["Grade"] = df["Total"].apply(get_grade)

print(df)
```
