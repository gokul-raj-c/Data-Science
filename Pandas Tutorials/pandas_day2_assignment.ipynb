3.	Employee Performance and Salary Analysis.

```python
employee={
    "name":['Alice','Bob','Charlie','David','Emma','Farah','George','Henry','Ishita','Jacob','Kalpana','Louis','Meena','Naveen','Oscar','Priya'],
    "department":['HR','IT','Finance','IT','Marketing','Engineering','Support','Sales','IT','Engineering','HR','Finance','Support','Engineering','IT','Marketing'],
    "age":[28,32,26,29,31,35,25,27,34,45,30,29,26,33,28,27],
    "experience":[5,8,3,6,7,12,2,3,10,20,6,5,4,9,5,3],
    "salary":[45000,85000,48000,90000,50000,120000,40000,42000,150000,180000,47000,46000,44000,95000,60000,100000],
    "score":[3.9,4.1,3.6,4.2,3.8,4.5,3.4,3.5,4.8,2.1,3.9,4.0,3.7,4.3,4.0,1.5]
}
```

i.	Create a data frame based on the above dataset.

```python
employee_df=pd.DataFrame(employee)
print(employee_df)
```

ii.	Check for the dimension of the dataset.

```python
print(employee_df.shape)
```

iii.	Take a statistical summary of the dataset.

```python
print(employee_df.describe())
```

```python
print(employee_df.describe(include='all'))
```

iv.	Use .iloc() to locate the 'Name' and 'Salary' columns in the DataFrame and check who has the highest salary.

```python
print(employee_df.iloc[:,[0,4]])
```

```python
print(employee_df.iloc[employee_df["salary"].idxmax()])
```

v.	Find the average monthly salary of each department and identify which department pays the most.

```python
avg_salary = employee_df.groupby("department")["salary"].mean()

print(avg_salary)
```

vi.	Identify outliers in the performance scores by visualizing them with a box plot.

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.boxplot(y=employee_df["score"])
plt.title("Performance Score Box Plot")
plt.ylabel("Performance Score")
plt.show()
```

vii.	Find Q1, Q3 and IQR.

```python
q1=employee_df['score'].quantile(0.25)
q3=employee_df['score'].quantile(0.75)
iqr=q3-q1
print(q1)
print(q3)
print(iqr)
```

viii.	Find the lower bound and upper bound using IQR.

```python
lower_bound=q1-1.5*iqr
upper_bound=q1+1.5*iqr
print(lower_bound)
print(upper_bound)
```

ix.	Remove the outliers present using IQR method.

```python
employee_df=employee_df[(employee_df["score"] >= lower_bound) & (employee_df["score"] <= upper_bound)]
print(employee_df)
```
