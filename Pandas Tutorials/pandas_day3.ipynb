```python
import pandas as pd
import numpy as np
```

Reindexing

```python
N=20
df=pd.DataFrame({
    'A':pd.date_range(start='2026-01-01',periods=N,freq='D'),
    'x':np.linspace(0,stop=N-1,num=N),
    'y':np.random.rand(N),
    'C':np.random.choice(['low','medium','high'],N),
    'D':np.random.normal(100,10,size=(N))
})
print(df)
```

Iteration of columns and rows

```python
for col in df:
  print(col)
```

1.iterrows()

```python
for row_index,row in df.iterrows():
  print("Row Index:",row_index)
  print(row)
```

2.itertuples()

```python
for row in df.itertuples():
  print(row)
```

Sorting

```python
unsorted_df=pd.DataFrame(np.random.randn(10,2),index=[1,4,6,2,3,5,9,8,0,7],columns=['col1','col2'])
print(unsorted_df)
```

By Label

```python
sorted_df=unsorted_df.sort_index()
print(sorted_df)
```

Order of Sorting

```python
sorted_df = unsorted_df.sort_index(ascending=False)
print(sorted_df)
```

Sort the Columns

```python
sorted_df = unsorted_df.sort_index(axis=1)
print(sorted_df)
```

By Value

```python
sorted_df = unsorted_df.sort_values(by='col1')
print(sorted_df)
```

'by' argument takes a list of column values.

```python
sorted_df = unsorted_df.sort_values(by=['col1','col2'])
print(sorted_df)
```

Sorting Algorithm

```python
unsorted_df=pd.DataFrame({'col1':[2,1,1,1],'col2':[1,3,2,4]})
print(unsorted_df)
sorted_df=unsorted_df.sort_values(by='col1',kind='mergesort')
print(sorted_df)
```

Working with Text Data

```python
s = pd.Series(['Tom', 'William Rick', 'John','Alber@t', np.nan, '1234','SteveSmith'])
```

1. lower()

```python
print(s.str.lower())
```

2. upper()

```python
print (s.str.upper())
```

3. len()

```python
print(s.str.len())
```

4. strip()

```python
print(s)
print ("After Stripping:")
print(s.str.strip())
```

5. split()

```python
print(s.str.split(' '))
```

6. contains()

```python
print(s.str.contains(' '))
```

7. replace()

```python
print(s.str.replace('@','$'))
```

8. count()

```python
print(s.str.count('m'))
```

9. startswith()

```python
print(s.str. startswith ('T'))
```

10. endswith()

```python
print(s.str.endswith('t'))
```

11. find()

```python
print(s.str.find('e'))
```

12. swapcase()

```python
print(s.str.swapcase())
```

13. islower()

```python
print(s.str.islower())
```

14. isupper()

```python
print(s.str.isupper())
```
