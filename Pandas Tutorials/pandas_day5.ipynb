```python
import pandas as pd
```

Applying a function to all the values in a column apply()  

```python
df=pd.DataFrame({
    "A":[2,3],
    "B":[4,5],
    "C":[6,7]
})
df
```

Creating a new column D by adding existing columns B and C

```python
def foo(row): 
    return row["B"] + row["C"] 
df["D"] = df.apply(foo, axis=1)
df
```

Creating a new column D by multiplying 2 with column A

```python
def foo(row): 
    return(row*2) 
df["D"] = df["A"].apply(foo)
df
```

Apply and lambda

```python
df["D"] = df.apply(lambda row: row["B"] + row["C"], axis=1)
df
```

Creating a new column D by multiplying 2 with column A 

```python
df["D"] = df["A"].apply(lambda row: row*2)
df
```
