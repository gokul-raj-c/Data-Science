```python
import pandas as pd
```

```python
df=pd.read_csv('Datasets/H.csv')
df_copy=df.copy()
df_copy.head()
```

```python
df_copy.isnull().sum()
```

```python
df_copy.duplicated()
```

```python
df_copy.info()
```

```python
df1 = df_copy.groupby('Department')['Employee_Satisfaction_Score'].mean().reset_index().sort_values(by='Employee_Satisfaction_Score', ascending=False)

print(df1)
```

```python
df2 = df_copy[df_copy['Resigned'] == True].groupby('Education_Level')['Resigned'].count()

total = df_copy.groupby('Education_Level')['Resigned'].count()

df2 = (df2 / total * 100).reset_index(name='Resignation_Rate')

print(df2)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.barplot(x='Education_Level', y='Resigned', data=df2)

plt.title('Resignation Rate by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Resignation Rate')
plt.show()
```
