**DataFrame**

pd.DataFrame(data, index, columns, dtype, copy)

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame()
print (df)
```

1. Create a DataFrame from Lists

```python
data=[1,2,3,4,5]
df=pd.DataFrame(data)
print(df)
```

Create a dataframe with column index

```python
data=[['Alex',10],['Bob',12],['clarke',13]]
df=pd.DataFrame(data,columns=['Name','Age'])
print(df)
```

2. Create a DataFrame from Dict of ndarrays or Lists

```python
data={
    'Name':['Tom','Jack','Steve','Ricky'],
    'Age':[28,34,29,42]
}
df=pd.DataFrame(data)
print(df)
```

 create an indexed DataFrame with dictionary and lists

```python
name_list=['tom','jack','steve','ricky']
age_list=[28,34,29,42]
data={'Name':name_list,'Age':age_list}
df=pd.DataFrame(data,index=['rank1','rank2','rank3','rank4'])
print(df)
```

create an indexed DataFrame with dictionary and numpy array

```python
name_arr=np.array(['tom','jack','steve','ricky'])
age_arr=np.array([28,34,29,42])
data={'Name':name_arr,'Age':age_arr}
df=pd.DataFrame(data,index=['rank1','rank2','rank3','rank4'])
print(df)
```

3.Create a DataFrame from I-iSt of Dicts

```python
data=[{'a':1,'b':2},{'a':5,'b':10,'c':20}]
df=pd.DataFrame(data)
print(df)
```

4. To create a DataFrame by passing a list of dicti0'1aries and the row
indices

```python
data=[{'a':1,'b':2},{'a':5,'b':10,'c':20}]
df=pd.DataFrame(data,index=['first','second'])
print(df)
```

5. To create a DataFrame with a list of dictionaries, row indices, and
column indices.

```python
data=[{'a':1,'b':2},{'a':5,'b':10,'c':20}]
#with two column indices, values same as dictionary
df=pd.DataFrame(data,index=['first','second'],columns=['a','b'])
print(df)

#with two column indices, with one index with other name
df=pd.DataFrame(data,index=['first','second'],columns=['a','b1'])
print(df)
```

6. Create a DataFra from DiCtOf Series

```python
name_series=pd.Series(['tom','jack','steve'],index=['a','b','c'])
age_series=pd.Series([28,34,29,42],index=['a','b','c','d'])
data={'Name':name_series,'Age':age_series}
df=pd.DataFrame(data)
print(df)
```

column selection

```python
d={
    'one':pd.Series([1,2,3],index=['a','b','c']),
    'two':pd.Series([1,2,3,4],index=['a','b','c','d'])
}
df=pd.DataFrame(d)
print(df['one'])
print(df[['one','two']])
```

Column Addition

```python
df['three']=pd.Series([10,20,30],index=['a','b','c'])
print(df)
df['four']=df['one']+df['three']
print(df)
```

Column Deletion

```python
del df['four']
print(df)
df.pop('three')
print(df)
```
