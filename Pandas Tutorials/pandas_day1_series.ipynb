```python
import pandas as pd
```

**Series**

1. List

```python
mylist=[1,2,3,4]
ds=pd.Series(mylist)
print(ds)
```

2. Array


```python
import numpy as np
arr=np.array([1,2,3,4])
df=pd.Series(arr)
print(df)
```

3. Dict

```python
mydict={
    "a":1,
    "b":2,
    "c":3,
    "d":4
}
ds=pd.Series(mydict)
print(ds)
```

4. Scalar value or constant

```python
ds=pd.Series(3,index=['a','b','c'])
print(ds)
print(ds.values)
```

Accessing Data from Series with position

```python
ds=pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
print(ds[0])
#retrieve the first thre Genent
print(ds[:3])
#retrieve the last three element
print(ds[-3:])
```

Retrieve Data Using Label (Index)

```python
ds=pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
#retrieve a single element
print(ds['b'])
#retrieve multiple elements
print(ds[['a','c','d']])
```

Series Basic Functionality

```python
s=pd.Series(np.random.randn(4))
print(s)
```

```python
#return list Labels of the series.
print(s.axes)
```

```python
#return boolean value saying whether object empty or not
print(s.empty)
```

```python
#no of dimensions
print(s.ndim)
```

```python
#size of series
print(s.size)
```

```python
#values
print(s.values)
```

```python
#head and tail
print(s.head(2))
print(s.tail(2))
```
