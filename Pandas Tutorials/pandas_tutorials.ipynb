```python
data={
    'product':['Good Day','Dairy Milk','Snickers','Bourbon'],
    'unit_price':[40,30,20,20],
    'quantity':[2,1,3,2]
}
df=pd.DataFrame(data)
print(df)
```

```python
df['price']=df['unit_price']*df['quantity']
print(df)
```

```python
df['discount']=pd.Series([2,0,2,3])
print(df)
```

```python
amt=df['unit_price']-(df['unit_price']*(df['discount']/100))
df['final_amount']=amt*df['quantity']
print(df)
```

```python
final_sum=df['final_amount'].sum()
total_amount=final_sum+(final_sum*0.05)
print(df)
print('final amount: ',final_sum)
print('gst: 5%')
print('total: ',total_amount)
```

Removal of Outliers

data points that are significantly different from the rest of the
data in a dataset. These values are either much higher or much lower than
most of the other data points, and they can distort statistical analyses,
leading to misleading conclusions.


1. Visual Inspection (Boxplots and Scatterplots)

2. Remove Outliers sing IQR (Interquartile Range)

QI (First Quartile): 25th percentile

Q3 (Third Quartile): 75th percentile

IQR=Q3-Q1

Lower Bound = Q1 - 1.5 * IQR

Upper Bound = Q3 + 1.5 * IQR

Any data points below the Lower Bound or above the Upper Bound are
considered outliers.


```python
data={
    'col':[10,20,30,40,100,200,300,500,1000]
}
df=pd.DataFrame(data)
print(df)
```

```python
Q1=df['col'].quantile(0.25)
Q3=df['col'].quantile(0.75)
IQR=Q3-Q1
```

```python
lower_bound=Q1-1.5*IQR
upper_bound=Q1+1.5*IQR
```

```python
df_no_outliers=df[(df['col'] >= lower_bound) & (df['col'] <= upper_bound)]
print(df_no_outliers)
```
