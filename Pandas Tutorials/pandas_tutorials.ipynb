```python
import pandas as pd
import numpy as np
```

```python
data={
    'name':pd.Series(['Tom','James','Ricky','Vin','Steve','Smith','Jack']),
    'age':pd.Series([25,26,25,23,30,29,23]),
    'rating':pd.Series([4.23,3.24,3.98,2.56,3.2,4.6,3.8])
}
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
print(df.T)
```

Row Selection

```python
data={
    'one':pd.Series([1,2,3], index=['a','b','c']),
    'two':pd.Series([1,2,3,4], index=['a','b','c','d'])
}
```

```python
df=pd.DataFrame(data)
print(df)
```

Selection by Label

```python
print(df.loc[:,'one'])
```

```python
print(df.loc['a':'c','one'])
```

```python
print(df.loc[['a','c'],'one'])
```

Selection by integer location

```python
print(df.iloc[:,0])
```

```python
print(df.iloc[0:3,0])
```

```python
print(df.iloc[[0,2,3],0])
```

Pandas - Descriptive Statistics

```python
data={
    'name':pd.Series(['Tom','James','Ricky','Vin','Steve','Smith','Jack']),
    'age':pd.Series([25,26,25,23,30,29,23]),
    'rating':pd.Series([4.23,3.24,3.98,2.56,3.2,4.6,3.8])
}
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
print(df[['age','rating']].mean())
```

```python
print(df[['age','rating']].median())
```

```python
print(df[['age','rating']].mode())
```

```python
print(df[['age','rating']].var())
```

```python
print(df[['age','rating']].std())
```

```python
print(df[['age','rating']].min())
```

```python
print(df[['age','rating']].max())
```

```python
from scipy.stats import skew, kurtosis
```

```python
print(kurtosis(df[['age','rating']], fisher=True))
```

```python
print(df[['age','rating']].quantile(0.25))
```

```python
print(df[['age','rating']].quantile(0.5))
```

```python
print(df[['age','rating']].quantile(0.75))
```

```python
print(skew(df[['age','rating']]))
```
