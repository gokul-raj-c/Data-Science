```python
import pandas as pd
import numpy as np
```

Filter Pandas Dataframe with multiple conditions

```python
data={
    'name':['rachel','monica','phoebe','ross','chandler','joey'],
    'age':[30,35,37,33,34,30],
    'salary':[100000,93000,88000,120000,94000,95000],
    'job':['designer','chef','masus','palentology','it','artist']
}
data_df=pd.DataFrame(data)
print(data_df)
```

1. Filter Pandas Dataframe with multiple conditions Using loc

```python
#Filter Name and Job from data with salary greater than 100000 age less than 40 job starting with D with loc method
print(data_df.loc[(data_df['salary'] >= 100000) & (data_df['age']<40) & (data_df['job'].str.startswith('d')), ['name','job']])
```

2. Filter Pandas Dataframe Using NumPy

```python
#Filter Name and Job from data with salary greater than 100000 age less than 40 job starting with D with numpy where method
filtered_values=np.where((data_df['salary'] >= 100000) & (data_df['age']<40) & (data_df['job'].str.startswith('d')))
print(data_df.iloc[filtered_values])
```

3. Filter Pandas Dataframe Using Query (eval and query works only
with columns)

```python
#Filter Name and Job from data with salary greater than 100000 age less than 40 job starting with D with query method
print(data_df.query('salary >= 100000 & age < 40 & job.str.startswith("d").values'))
```

4. Pandas Boolean indexing multiple conditions standard way
("Boolean indexing" works with values in a column only)

```python
#Filter Name and Job from data with salary greater than 100000 age less than 40 job starting with D with pandas filtering
print(data_df[(data_df['salary'] >= 10000) & (data_df['age']<40) & (data_df['job'].str.startswith('d'))][['name','age','salary']])
```

5. Eval multiple conditions ("eval" and "query" works only with
columns )

```python
#Filter Name and Job from data with salary greater than 100000 age less than 40 job starting with D with eval method
print(data_df[data_df.eval('salary >= 100000 & age < 40 & job.str.startswith("d").values')])
```

eval method can also be used to perform arithmetic operations

```python
data_df['salary in thousands']=data_df.eval('salary/1000')
print(data_df)
```

6. How to Filter DataFrame by List of Values in Pandas

```python
#Filter Name and Job from data using isin method
job_list=['designer','it']
filtered_df=data_df[data_df['job'].isin(job_list)]
print(filtered_df)
```

7. How to Filter DataFrame using filter in Pandas

```python
#Filter Name and Job from data using filter method
filtered_df=data_df.filter(items=['name','age'])
print(filtered_df)
```

8. Selecting Columns usinu filter and like

```python
#Filter columns containing 'a' from data using filter method and like
filtered_df=data_df.filter(like='a', axis=1)
print(filtered_df)
```

9. Selecting Columns Using regex

```python
#Filter columns starting with A from data using filter method and regex
filtered_df=data_df.filter(regex='^a', axis=1)
print(filtered_df)
```

Merging/Joining

```python
left=pd.DataFrame({
    'id':[1,2,3,4,5],
    'name':['alex','amy','allen','alice','ayoung'],
    'subject_id':['sub1','sub2','sub3','sub4','sub5']
})
right=pd.DataFrame({
    'id':[1,2,3,4,5],
    'name':['billy','brian','bran','bryce','betty'],
    'subject_id':['sub2','sub4','sub3','sub6','sub5']
})
```

```python
print(left)
print(right)
```

Merge Two DataFrames o a Key

```python
print(pd.merge(left, right,on='id'))
```

Merge Two DataFrames on Multiple Keys

```python
print(pd.merge(left,right,on=['id','subject_id']))
```

Merge Using 'how' Argument

Left Join

```python
print(pd.merge(left,right,on='subject_id',how='left'))
```

Right Join

```python
print(pd.merge(left,right,on='subject_id',how='right'))
```

Outer Join

```python
print(pd.merge(left,right,on='subject_id',how='outer'))
```

Inner Join

```python
print(pd.merge(left,right,on='subject_id',how='inner'))
```

Dataframe Concatenation

```python
one=pd.DataFrame({
    'name':['alex','amy','allen','alice','ayoung'],
    'subject_id':['sub1','sub2','sub3','sub4','sub5'],
    'marks_scored':[98,90,87,69,78]
},index=[1,2,3,4,5])
two=pd.DataFrame({
    'name':['billy','brian','bran','bryce','betty'],
    'subject_id':['sub2','sub4','sub3','sub6','sub5'],
    'marks_scored':[89,80,79,97,88]
},index=[1,2,3,4,5])
```

```python
print(one)
print(two)
```

Concatenating Objects

```python
print(pd.concat([one,two]))
```

```python
print(pd.concat([one,two],axis=1) )
```
