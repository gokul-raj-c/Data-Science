```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

**Plot from a DataFrame**

Plot a Scatter Diagram using Pandas

```python
data={
    'unemployment_rate':[6.1, 5.8, 5.7, 5.7, 5.8, 5.6, 5.5, 5.3, 5.2, 5.2],
    'stock_price': [1500, 1520, 1525, 1523, 1515, 1540, 1545, 1560, 1555, 1565]
}
df=pd.DataFrame(data)
print(df)
```

```python
df.plot(x='unemployment_rate',y='stock_price',kind='scatter')
plt.show()
```

Plot a Line Chart using Pandas

```python
data={
    'year':[1920,1930,1940,1950,1960,1970,1980,1990,2000,2010],
    'unemployment_rate':[9.8,12,8,7.2,6.9,7,6.5,6.2,5.5,6.3]
}
df=pd.DataFrame(data)
print(df)
```

```python
df.plot(x='year', y='unemployment_rate',kind='line')
plt.show()
```

Plot a Bar Chart using Pandas

```python
data={
    'country':['USA','Canada','Germany','UK','France'],
    'gdp':[45000,42000,52000,49000,47000]
}
df=pd.DataFrame(data)
print(df)
```

```python
df.plot(x='country', y='gdp', kind='bar')
plt.show()
```

Plot a Pie Chart using Pandas

```python
data={
    'tasks':[300,500,700]
}
df=pd.DataFrame(data, index=['pending','ongoing','completed'])
print(df)
```

```python
df.plot.pie(y='tasks',figsize=(5, 5),autopct='%1.1f%%', startangle=90)
plt.show()
```

Add a Timedelta to a Date Column 

```python
df = pd.DataFrame({ 
'event_time': pd.date_range('2024-01-01 08:00:00', 
periods=5, freq='D') 
})
print(df)
```

```python
# Define a timedelta 
delta = pd.Timedelta('2 days 2 hours 15 minutes 30 seconds') 
print(delta)
```

```python
# Create a new column with the adjusted time 
df['event_end'] = df['event_time'] + delta
print(df)
```

Combine Duration Columns into a Timedelta 

```python
# Create random duration components 
np.random.seed(42)
df=pd.DataFrame({
    'days':np.random.randint(0,5,size=10),
    'hours':np.random.randint(0,24,size=10),
    'minutes':np.random.randint(0,60,size=10),
    'seconds':np.random.randint(0,60,size=10)
})
print(df)
```

```python
# Combine into a Timedelta column 
df['duration']=pd.to_timedelta(
    df['days'].astype(str)+ ' days '+
    df['hours'].astype(str)+ ' hours '+
    df['minutes'].astype(str)+ ' minutes '+
    df['seconds'].astype(str)+ ' seconds '
)
print(df)
```

Filter data based on Timedelta

```python
np.random.seed(42) 
df = pd.DataFrame({ 
    'days': np.random.randint(0, 5, size=10), 
    'hours': np.random.randint(0, 24, size=10), 
    'minutes': np.random.randint(0, 60, size=10), 
    'seconds': np.random.randint(0, 60, size=10) 
    })
print(df)
```

```python
df['duration'] = pd.to_timedelta( 
df['days'].astype(str) + ' days ' + 
df['hours'].astype(str) + ' hours ' + 
df['minutes'].astype(str) + ' minutes ' + 
df['seconds'].astype(str) + ' seconds' 
)
print(df)
```

```python
filtered_df = df[df['duration'] > pd.Timedelta('3 days 6 hours')] 
print(filtered_df) 
```

```python
#Filter between 1 and 3 days
filtered_range = df[(df['duration'] >= pd.Timedelta('1 days')) & (df['duration'] <= pd.Timedelta('3 days'))] 
print(filtered_range)
```

Addition Operations 

```python
s = pd.Series(pd.date_range('2012-1-1', periods=3, freq='D')) 
td = pd.Series([ pd.Timedelta(days=i) for i in range(3) ]) 
df = pd.DataFrame(dict(A = s, B = td)) 
df['C']=df['A']+df['B'] 
print(df)
```

Subtraction Operation 

```python
df['C']=df['A']-df['B'] 
df['D']=df['C']-df['B'] 
print(df)
```
