```python
import pandas as pd
```

```python
df=pd.read_csv("Superstore.csv", encoding='cp437')
print(df.head())
```

Analyzing the columns and datatypes in the dataset

```python
df.info()
```

```python
df.columns=df.columns.str.replace(' ','_')
```

```python
df.info()
```

Change data type

```python
print(df.dtypes)
```

```python
#change dtype to date time
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'])
```

```python
print(df.dtypes)
```

Convert sales and profit to apprx 2 decimals

```python
df['Sales']
```

```python
df[['Sales', 'Profit']] = df[['Sales', 'Profit']].round(2)
```

```python
df[['Sales', 'Profit']]
```

values inside a particular column from space to underscore

```python
df['Ship_Mode']
```

```python
df['Ship_Mode'] = df['Ship_Mode'].str.replace(' ', '_')
```

```python
df['Ship_Mode']
```

Values inside customer name, split it to first name and last name

```python
df[['First_Name', 'Last_Name']] = df['Customer_Name'].str.split(' ', n=1, expand=True)
```

```python
df[['Customer_Name','First_Name', 'Last_Name']]
```

Remove columns

```python
df.drop(['First Name', 'Last Name'], axis=1, inplace=True)
```

```python
df.head()
```

```python
df.groupby('Ship_Mode')['Sales'].sum().sort_values(ascending=False)
```
